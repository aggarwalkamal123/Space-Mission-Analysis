```python
# ==========================================
# Function to Extract Country from Location
# ==========================================

def extract_country(location):

    # Convert location to string
    location = str(location).strip()

    # Special cases where the location does not end with a country
    if 'Pacific Ocean' in location:
        return 'USA'

    elif 'Pacific Missile Range Facility' in location:
        return 'USA'

    elif 'Yellow Sea' in location:
        return 'South Korea'

    elif 'Barents Sea' in location:
        return 'Russia'

    elif 'Shahrud Missile Test Site' in location:
        return 'Iran'

    # For normal locations, country is the last comma-separated value
    else:
        return location.split(',')[-1].strip()
```

```python
# ==========================================
# Extract Country from Launch Location
# ==========================================

df_clean['Country'] = df_clean['Location'].apply(
    extract_country
)

df_clean[['Location', 'Country']].head(10)
```
